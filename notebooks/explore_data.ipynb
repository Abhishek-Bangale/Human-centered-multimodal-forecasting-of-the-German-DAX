```python
import pandas as pd
import os
import csv # For quoting constant

# --- Configuration ---
PARQUET_INPUT_FILE = "processed_gdelt_leads/dax40_gdelt_article_leads.parquet"
CSV_OUTPUT_FILE = "processed_gdelt_leads/dax40_gdelt_article_leads.csv"
# --- End Configuration ---

def main_convert_parquet_to_csv():
    if not os.path.exists(PARQUET_INPUT_FILE):
        print(f"Error: Parquet input file not found: {PARQUET_INPUT_FILE}")
        return

    print(f"Loading Parquet file: {PARQUET_INPUT_FILE} ...")
    try:
        df = pd.read_parquet(PARQUET_INPUT_FILE)
        print(f"Successfully loaded {len(df)} rows from Parquet.")
    except Exception as e:
        print(f"Error loading Parquet file: {e}")
        return

    print(f"Attempting to save to CSV: {CSV_OUTPUT_FILE} ...")
    try:
        # Crucial parameters for robust CSV writing:
        # - index=False: Don't write the DataFrame index as a column
        # - encoding='utf-8-sig': 'utf-8-sig' writes a BOM (Byte Order Mark), 
        #   which helps some programs (like Excel) correctly recognize UTF-8.
        #   'utf-8' is also fine.
        # - quoting=csv.QUOTE_MINIMAL: Only quote fields that contain the separator, quote char, or newlines.
        #   csv.QUOTE_ALL will quote every field, making the file larger but safer.
        # - escapechar: If needed, but pandas handles quoting well.
        df.to_csv(CSV_OUTPUT_FILE, index=False, encoding='utf-8-sig', quoting=csv.QUOTE_MINIMAL)
        print(f"Successfully saved data to CSV: {CSV_OUTPUT_FILE}")
        print(f"CSV file size: {os.path.getsize(CSV_OUTPUT_FILE) / (1024*1024):.2f} MB")

    except UnicodeEncodeError as uee:
        print(f"UnicodeEncodeError: Could not encode some characters to CSV with 'utf-8-sig'. {uee}")
        print("This often happens with exotic characters. You might need to preprocess text further or try 'utf-8'.")
    except Exception as e:
        print(f"Error saving data to CSV: {type(e).__name__} - {e}")

if __name__ == '__main__':
    main_convert_parquet_to_csv()
```
